## 1. List and Dictionary Comprehensions

**What is it?**
Comprehensions provide a concise way to create lists or dictionaries from existing iterables without writing verbose `for` loops.

**When to use it:**
Use them when transforming, filtering, or mapping data elements dynamically in a single, highly readable line of code.


List Comprehension

````python
# Traditional For Loop
squares = []
for x in range(5):
    if x % 2 == 0:
        squares.append(x ** 2)


# List Comprehension
squares = [x ** 2 for x in range(10) if x % 2 == 0]
# Output: [0, 4, 16]
print(squares)
````

````python
ls = [x + 5 * 7 ** 2 for x in range(10) if x % 9 == 0 if x % 5]
print(ls)
````

````python
# Returns: [0, "I don't know", 4, "I don't know", 16]
some_expression = [x ** 2 if x % 2 == 0 else "I don't know" for x in range(5)]
````

* **Filtering (No else allowed):**  
  `[expression for item in iterable if condition]` $\rightarrow$ Elements are only included if the condition evaluates to `True`.

* **Conditional Transformation:**  
  `[expression_if_true if condition else expression_if_false for item in iterable]` $\rightarrow$ Transforms elements conditionally using inline `if-else` logic.

````python
colors = ["red", "green"]
objects = ["hat", "shirt", "shoes"]

# Pair them up, but skip red shoes
outfits = [f"{c} {o}" for c in colors for o in objects if not (c == "red" and o == "shoes")]

print(outfits)
# Output: ['red hat', 'red shirt', 'green hat', 'green shirt', 'green shoes']
````

````python
# Keep numbers that are divisible by 2 AND divisible by 3
numbers = [x for x in range(20) if x % 2 == 0 if x % 3 == 0]

print(numbers)
# Output: [0, 6, 12, 18]
````

````python
# Label numbers as Even or Odd
labels = ["Even" if x % 2 == 0 else "Odd" for x in range(5)]
# Output: ['Even', 'Odd', 'Even', 'Odd', 'Even']

# Works identically in dictionaries
dict_labels = {x: "Even" if x % 2 == 0 else "Odd" for x in range(3)}
# Output: {0: 'Even', 1: 'Odd', 2: 'Even'}
````

## 2. Lambda Functions

**What is a Lambda?**
A lambda function is an anonymous, single-line function defined without a name. It is written using the `lambda` keyword.

**Syntax:**
`lambda arguments: expression`

**When to use it:**
Use lambdas for quick, throwaway operations where defining a standard function with `def` is overkill—especially as helper functions inside methods like `map()`, `filter()`, or pandas `.apply()`.


Lambda Function with map, filter, reduce
# Lambda function
```lambda arguments: expression```

# map
```map(lambda argument: transformation_expression, iterable)```

#filter
```filter(lambda argument: boolean_condition_expression, iterable)```


### Lambda Use Cases & Examples

Below are standard examples demonstrating how anonymous lambda functions operate in Python. We will explore:
1. **Basic Lambda Functions**
2. **Pass multiple arguments**
3. **Combining with `map()`**
4. **Combining with `filter()`**
5. **Combining with `sorted()` using custom sorting keys**

````python
# 1. Basic Single-Argument Lambda
double = lambda x: x * 2
print("Double of 5:", double(5))  # Output: 10

# 2. Multiple arguments
add = lambda a, b: a + b
print("Sum of 3 and 4:", add(3, 4))  # Output: 7

# 3. Used with map() (transforming items)
numbers_list = [1, 2, 3, 4]
squared = list(map(lambda x: x ** 2, numbers_list))
print("Squared numbers:", squared)  # Output: [1, 4, 9, 16]

# 4. Used with filter() (extracting items)
evens = list(filter(lambda x: x % 2 == 0, numbers_list))
print("Even numbers:", evens)  # Output: [2, 4]

# 5. Used with sorted() (custom sorting key)
students = [("Alice", 25), ("Bob", 20), ("Charlie", 23)]
students.sort(key=lambda student: student[1])
print("Sorted by age:", students)  # Output: [('Bob', 20), ('Charlie', 23), ('Alice', 25)]

````

````python
ls = [ [5,6,7], [1,2,3], [3,4,5]]
ls.sort(key =lambda x : (x[0], x[1]))
print(ls)
````

## 3. Map, Filter, and Reduce

These functional programming pillars are highly optimized for processing elements in collections without manual loops.

*   **`map()`**: Transforms every single element in an iterable using a function.
*   **`filter()`**: Extracts only the elements of an iterable that satisfy a truth condition.
*   **`reduce()`**: Iteratively applies a function to elements to accumulate them into a single summary value.



#reduce

from functools import reduce
reduce(function, iterable[, initializer])


```expression```


How It Works (Step-by-Step)

If you pass a lambda function like lambda x, y: x + y and a list [1, 2, 3, 4]:
1. It takes the first two elements (1 and 2), adds them, and returns 3.
2. It takes that result (3) and the next element (3), adds them, and returns 6.
3. It takes that result (6) and the last element (4), adds them, and returns 10.

````python
from functools import reduce

numbers = [1, 2, 3, 4, 5]
total = reduce(lambda x, y: x + y, numbers)
print(total)  # Output: 15


# Starts the sum at 100 instead of the first element
total = reduce(lambda x, y: x * y, numbers, 100)
print(total)  # Output: 106
````

````python
numbers = [23, 76, 12, 89, 45]
highest = reduce(lambda a, b: a if a > b else b, numbers)
print(highest)  # Output: 89


````

#There is more to data manipulation

```df.groupby('category_col').agg({'salary': 'mean', 'age': 'median'})```

```pd.merge(df1, df2, on='id_column', how='inner')```

```pd.concat([df1, df2], axis=0)```




````python
# Ensure the dataset is loaded
housing_df = pd.read_csv('/content/sample_data/california_housing_train.csv')

# Create a simple category for demonstration
housing_df['housing_age_category'] = pd.cut(housing_df['housing_median_age'],
                                            bins=[0, 20, 40, 100],
                                            labels=['New', 'Moderate', 'Old'])

# 1. Basic groupby: Find mean house value for each age category
print("--- AVERAGE VALUE BY AGE CATEGORY ---")
avg_value = housing_df.groupby('housing_age_category', observed=False)['median_house_value'].mean()
print(avg_value)

# 2. Multi-column aggregation
print("\n--- MULTI-COLUMN AGGREGATION ---")
agg_summary = housing_df.groupby('housing_age_category', observed=False).agg({
    'total_rooms': 'mean',
    'median_income': 'median'
})
display(agg_summary)
````

# Outlier detection

````python
# Outlier detection with the Interquartile Range (IQR) method
q1 = housing_df['median_house_value'].quantile(0.25)
q3 = housing_df['median_house_value'].quantile(0.75)

iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

housing_no_outliers = housing_df[(housing_df['median_house_value'] >= lower_bound) & (housing_df['median_house_value'] <= upper_bound)]
print(f"Dataset size before: {len(housing_df)} | After: {len(housing_no_outliers)}")
````

## 1. The Definitive Chart Blueprint

| Chart & Graph | Why It Exists (Purpose) | Kind of Data It Requires | Shortest Pandas Syntax |
| :--- | :--- | :--- | :--- |
| **Line Chart** | Tracks trends, changes, and movements over continuous intervals. | **X:** Time/Dates or ordered numbers<br>**Y:** Continuous numeric | `df.plot(x='Date', y='Value')` |
| **Vertical Bar** | Compares individual values or quantities across discrete categories. | **X:** Categorical/Text labels<br>**Y:** Numeric amounts | `df.plot(kind='bar')` |
| **Horizontal Bar** | Same as vertical bar, but highly readable for long text labels. | **X:** Numeric amounts<br>**Y:** Categorical/Text labels | `df.plot(kind='barh')` |
| **Histogram** | Shows the frequency distribution, spread, and shape of a single dataset. | **X:** Continuous numeric (binned)<br>**Y:** Frequency count | `df.plot(kind='hist', bins=10)` |
| **Density (KDE)** | A smooth, continuous curve estimation of a histogram's shape. | **X:** Continuous numeric<br>**Y:** Probability density | `df.plot(kind='kde')` |
| **Scatter Plot** | Detects relationships, clusters, correlations, or patterns between two metrics. | **X:** Numeric column<br>**Y:** Numeric column | `df.plot(kind='scatter', x='A', y='B')` |
| **Box Plot** | Shows data distribution quartiles, spread, skewness, and outliers. | **Columns:** Numeric values grouped by categories | `df.plot(kind='box')` |
| **Area Plot** | Tracks trends over time while illustrating proportional composition. | **X:** Time/Dates<br>**Y:** Multiple numeric stacked series | `df.plot(kind='area')` |
| **Pie Chart** | Shows part-to-whole relationships (best used with 5 or fewer categories). | **Index:** Categories<br>**Values:** Non-negative metrics | `df.plot(kind='pie', y='Value')` |

---

## 2. Core Data Classifications

Your chart choice is dictated by your data types. Classify columns first:

*   **🔢 Numerical Continuous:** Decimals, measurements, price, speed, temperature.
    *   *Best Charts:* Line, Histogram, Scatter, Box, Hexbin
*   **🎲 Numerical Discrete:** Whole integer counts, number of items, days.
    *   *Best Charts:* Bar, Histogram
*   **🏷️ Categorical Nominal:** Text labels with no inherent ranking (e.g., Countries, Colors, Departments).
    *   *Best Charts:* Bar, Pie
*   **🏅 Categorical Ordinal:** Text labels with a strict inherent order (e.g., Low/Medium/High, Ratings 1-5).
    *   *Best Charts:* Ordered Bar
*   **📅 Time-Series:** Dates, timestamps.
    *   *Best Charts:* Line, Area


````python
# 2. Distribution of a Single Metric: Histogram with KDE
# Use Case: Analyze the frequency distribution and density of housing ages
plt.figure(figsize=(10, 6))
sns.histplot(data=housing_df, x='housing_median_age', bins=30, kde=True, color='purple')
plt.title('Distribution of Housing Median Age')
plt.xlabel('Median Age of Houses')
plt.ylabel('Frequency')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
````

````python
# 3. Categorical comparison: Box Plot
# Use Case: Bin continuous house ages to see statistical quartiles and outlier distributions
housing_df['age_group'] = pd.cut(housing_df['housing_median_age'],
                                 bins=[0, 15, 30, 45, 100],
                                 labels=['New (<15y)', 'Moderate (15-30y)', 'Old (30-45y)', 'Historical (45y+)'])

plt.figure(figsize=(10, 6))
sns.boxplot(data=housing_df, x='age_group', y='median_house_value', hue='age_group', palette='coolwarm', legend=False)
plt.title('House Value Distribution Across Age Categories')
plt.xlabel('Age Group Category')
plt.ylabel('Median House Value (USD)')
plt.grid(True, linestyle='--', alpha=0.3)
plt.show()
````



## Working with External Libraries (AI, ML & Deep Learning Ecosystem)

In the real world, we do not build machine learning models or neural networks from scratch. Instead, we stand on the shoulders of optimized, industry-standard libraries. Here is a curated guide to the essential packages in the modern AI ecosystem:

### 1. Data Manipulation & Array Computation
*   **Pandas (`pd`):** The primary library for structured tabular data (DataFrames). Used for reading, cleaning, joining, and filtering datasets.
*   **NumPy (`np`):** Provides support for large, multi-dimensional arrays and matrices, along with a collection of high-level mathematical functions to operate on these arrays.

### 2. Data Visualization
*   **Matplotlib (`plt`):** The foundation of plotting in Python. Highly customizable, allowing you to build line charts, histograms, scatter plots, and more.
*   **Seaborn (`sns`):** Built on top of Matplotlib, Seaborn provides a high-level interface for drawing attractive, statistically informative graphics with less boilerplate code.

### 3. Classical Machine Learning
*   **Scikit-Learn (`sklearn`):** The premier library for classical Machine Learning algorithms. It contains simple and efficient tools for predictive data analysis, including:
    *   **Preprocessing:** Scaling, encoding categorical labels, splitting training/testing sets.
    *   **Supervised Models:** Linear Regression, Logistic Regression, Decision Trees, Random Forests, Support Vector Machines (SVM).
    *   **Evaluation Metrics:** Confusion matrices, accuracy, precision, recall, and F1-scores.

### 4. Deep Learning & Artificial Neural Networks
*   **TensorFlow:** An open-source end-to-end platform for machine learning developed by Google, optimized for high-performance scale and production serving.
*   **Keras:** A high-level, user-friendly API designed for human beings to build and train deep learning models rapidly. Keras runs directly on top of TensorFlow, making it incredibly popular for both beginners and developers.
